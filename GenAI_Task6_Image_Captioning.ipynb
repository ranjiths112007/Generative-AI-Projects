```python
# Task 6 - Image Captioning with BLIP
# pip install transformers torch pillow requests
import torch
import requests
from PIL import Image
from io import BytesIO
from transformers import BlipProcessor, BlipForConditionalGeneration
```

```python
MODEL_NAME = 'Salesforce/blip-image-captioning-base'
processor = BlipProcessor.from_pretrained(MODEL_NAME)
model = BlipForConditionalGeneration.from_pretrained(MODEL_NAME)
print('BLIP image-captioning model loaded successfully.')
```

```python
# Public sample image used for the demonstration
image_url = 'https://images.unsplash.com/photo-1517248135467-4c7edcad34c4?w=800'
image = Image.open(BytesIO(requests.get(image_url, timeout=30).content)).convert('RGB')
display(image)
```

```python
inputs = processor(images=image, return_tensors='pt')
with torch.no_grad():
    output_ids = model.generate(**inputs, max_new_tokens=30)
caption = processor.decode(output_ids[0], skip_special_tokens=True)
print('Generated Caption:')
print(caption)
```

```python
# Captioning can also be guided with a text prompt.
prompt = 'a photo of'
prompt_inputs = processor(images=image, text=prompt, return_tensors='pt')
with torch.no_grad():
    prompted_ids = model.generate(**prompt_inputs, max_new_tokens=30)
prompted_caption = processor.decode(prompted_ids[0], skip_special_tokens=True)
print('Prompt-guided Caption:')
print(prompted_caption)
```
