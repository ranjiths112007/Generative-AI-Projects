```python
import transformers
import huggingface_hub
from transformers import pipeline
from transformers.pipelines import PIPELINE_REGISTRY
print("Transformers:", transformers.__version__)
print("Hugging Face Hub:", huggingface_hub.__version__)
```

```python
print("Transformers version:", transformers.__version__)
print("Summarization available:",
      "summarization" in PIPELINE_REGISTRY.get_supported_tasks())
```

```python
text = """
Artificial intelligence has become one of the most influential technologies
in the modern world. It is being used in many areas such as healthcare,
education, transportation, banking, agriculture, and entertainment.
Artificial intelligence enables machines to perform tasks that normally
require human intelligence, including learning, problem-solving,
decision-making, and understanding language. In healthcare, AI systems can
assist doctors in identifying diseases and analyzing medical images. In
education, AI-powered applications can provide personalized learning
experiences based on the needs of individual students. Businesses use AI to
analyze large amounts of data, understand customer behavior, and improve
their services. Self-driving vehicles also use artificial intelligence to
identify objects, understand road conditions, and make driving decisions.
Despite these advantages, artificial intelligence also creates several
challenges. Issues such as data privacy, algorithmic bias, job displacement,
security, and the responsible use of AI have become important concerns.
Developers and organizations must ensure that artificial intelligence
systems are transparent, fair, reliable, and safe. Governments and
researchers are also working to create guidelines for the responsible
development of AI technologies. As artificial intelligence continues to
advance, it is expected to become even more integrated into everyday life.
Therefore, understanding both the benefits and risks of AI is important for
ensuring that the technology contributes positively to society."""
```

```python
print("Loading model...")

summarizer = pipeline(
    "summarization",
    model="facebook/bart-large-cnn"
)

print("Model loaded successfully!")
summary_result = summarizer(
    text,
    max_length=80,
    min_length=30,
    do_sample=False
)

summary = summary_result[0]["summary_text"]

print("Generated Summary:")
print(summary)

```

```python
original_word_count = len(text.split())
summary_word_count = len(summary.split())

print(f"The original paragraph contains {original_word_count} words.")
print("Summary Word Count  :", summary_word_count)
```
